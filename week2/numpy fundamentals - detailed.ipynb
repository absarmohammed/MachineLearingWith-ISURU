### 2: PYTHON LISTS VS NUMPY ARRAYS

#### 2.1 Memory usage comparison

```python
# Create equivalent data structures
size = 10000
python_list = list(range(size))
numpy_array = np.arange(size)
```

```python
# Calculate memory usage
list_memory = sys.getsizeof(python_list) + sum(sys.getsizeof(i) for i in python_list)
array_memory = numpy_array.nbytes + numpy_array.__sizeof__()
```

```python
print(f"Data size: {size:,} integers")
print(f"Python list memory: {list_memory:,} bytes ({list_memory/1024/1024:.2f} MB)")
print(f"NumPy array memory: {array_memory:,} bytes ({array_memory/1024/1024:.2f} MB)")
print(f"Memory reduction: {list_memory/array_memory:.1f}x less memory with NumPy")
print(f"Memory savings: {(list_memory - array_memory):,} bytes")
```

NumPy consumes less resources because it stores data in contiguous memory blocks with fixed, low-overhead data types, unlike Python lists which store pointers to individual Python objects with extra memory overhead. This makes NumPy arrays more compact and faster for numerical operations.

#### 2.2 Basic operations performance comparison

```python
# Test different sizes to show scalability
test_sizes = [1000, 10000, 100000, 500000]
```

```python
print("Size\t\tList Time\tNumPy Time\tSpeedup")
print("-" * 55)
for size in test_sizes:
    # Create test data
    list_a = list(range(size))
    list_b = list(range(size, 2*size))
    array_a = np.arange(size)
    array_b = np.arange(size, 2*size)
    
    # Time list addition
    start_time = time.time()
    list_result = [a + b for a, b in zip(list_a, list_b)]
    list_time = time.time() - start_time
    
    # Time numpy addition
    start_time = time.time()
    array_result = array_a + array_b
    numpy_time = time.time() - start_time
    
    speedup = list_time / numpy_time if numpy_time > 0 else float('inf')
    print(f"{size:,}\t\t{list_time:.6f}s\t{numpy_time:.6f}s\t{speedup:.1f}x")
```

#### 2.3 Complex mathematical operations

```python
size = 100000
print(f"\nTesting with {size:,} elements:")

# Create test data
python_list = [float(i) for i in range(size)]
numpy_array = np.arange(size, dtype=float)

print("\n📊 Operation Comparison:")
operations = [
    ("Element-wise multiplication", lambda x: [i * 2.5 for i in x], lambda x: x * 2.5),
    ("Square root", lambda x: [i**0.5 for i in x], lambda x: np.sqrt(x)),
    ("Exponential", lambda x: [2.71828**i for i in x[:100]], lambda x: np.exp(x[:100])),  # Much smaller size for exp
    ("Trigonometric (sin)", lambda x: [np.sin(i) for i in x[:10000]], lambda x: np.sin(x[:10000]))  # Smaller size
]

for op_name, list_op, numpy_op in operations:
    try:
        # Time list operation
        start_time = time.time()
        list_result = list_op(python_list)
        list_time = time.time() - start_time
        
        # Time numpy operation  
        start_time = time.time()
        numpy_result = numpy_op(numpy_array)
        numpy_time = time.time() - start_time
        
        speedup = list_time / numpy_time if numpy_time > 0 else float('inf')
        print(f"  {op_name:25s}: List={list_time:.6f}s, NumPy={numpy_time:.6f}s, Speedup={speedup:.1f}x")
    except OverflowError:
        print(f"  {op_name:25s}: Operation resulted in overflow - values too large to compute")
```

#### 2.4 Matrix operations: The real power of NumPy

##### 2.4.1 Matrix creation comparison

```python
matrix_size = 1000

# Python list approach (nested lists)
start_time = time.time()
python_matrix = [[i*j for j in range(matrix_size)] for i in range(matrix_size)]
list_creation_time = time.time() - start_time

# NumPy approach
start_time = time.time()
numpy_matrix = np.arange(matrix_size**2).reshape(matrix_size, matrix_size)
numpy_creation_time = time.time() - start_time

print(f"  {matrix_size}x{matrix_size} matrix creation:")
print(f"  Python lists: {list_creation_time:.6f}s")
print(f"  NumPy arrays: {numpy_creation_time:.6f}s")
print(f"  Creation speedup: {list_creation_time/numpy_creation_time:.1f}x faster")
```

##### 2.4.2 Matrix multiplication: The ultimate test

```python
matrix_sizes = [50, 100, 200, 300]
print("Size\t\tPython Lists\tNumPy Arrays\tSpeedup\t\tComplexity")
print("-" * 70)

for size in matrix_sizes:
    # Create matrices
    # Python: nested lists
    matrix_a_list = [[np.random.rand() for _ in range(size)] for _ in range(size)]
    matrix_b_list = [[np.random.rand() for _ in range(size)] for _ in range(size)]
    
    # NumPy: arrays
    matrix_a_np = np.random.rand(size, size)
    matrix_b_np = np.random.rand(size, size)
    
    # Python list matrix multiplication (manual implementation)
    start_time = time.time()
    result_list = [[sum(matrix_a_list[i][k] * matrix_b_list[k][j] for k in range(size)) 
                   for j in range(size)] for i in range(size)]
    list_time = time.time() - start_time
    
    # NumPy matrix multiplication
    start_time = time.time()
    result_np = np.dot(matrix_a_np, matrix_b_np)
    numpy_time = time.time() - start_time
    
    speedup = list_time / numpy_time if numpy_time > 0 else float('inf')
    complexity = f"O(n³) = O({size**3:,})"
    
    print(f"{size}x{size}\t\t{list_time:.6f}s\t{numpy_time:.6f}s\t{speedup:.0f}x\t\t{complexity}")
```

### 2.5 Why NumPy is Faster: Technical Deep Dive

1. **C Implementation**:
   - NumPy core is written in C, not Python
   - Direct memory access without Python overhead
   - Optimized assembly instructions (BLAS/LAPACK)

2. **Vectorization**:
   - Single instruction operates on multiple data (SIMD)
   - CPU can process multiple elements simultaneously
   - Eliminates Python's interpreted loop overhead

3. **Memory Layout**:
   - Contiguous memory allocation
   - Better cache locality and memory access patterns
   - Homogeneous data types (no type checking overhead)

4. **Optimized Algorithms**:
   - Uses highly optimized mathematical libraries
   - Parallel processing capabilities
   - Advanced algorithms for complex operations

5. **Less Python Overhead**:
   - Minimal Python function call overhead
   - Direct memory manipulation
   - Compiled code execution vs interpreted code

### 3. Practical vectorization examples

#### 3.1 Vectorization & Broadcasting in NumPy

#### Vectorization
- Allows operations on entire arrays without explicit loops
- Performs element-wise operations efficiently
- Leverages optimized C code and CPU SIMD instructions

#### Broadcasting
- Allows operations between arrays of different shapes
- Automatically expands smaller arrays to match larger ones
- Rules:
1. Arrays must have same number of dimensions, or
2. One array can have fewer dimensions
3. Size of each dimension must be equal or 1

The following cells demonstrate these concepts with a portfolio analysis example:
- Calculating portfolio values (vectorization)
- Applying price changes across different stocks (broadcasting)


```python
# Example 01
a = np.ones(6)
b = np.ones((6, 1))
c = np.ones((1, 6))

ab = a + b
ac = a + c
bc = b + c
```

#### 3.4 Practical broadcasting - portfolio rebalancing

```python
weights = np.array([0.4, 0.3, 0.2, 0.1])  # Portfolio weights
asset_returns = np.array([[0.08, 0.12, -0.02, 0.15],   # Asset 1 quarterly returns
                         [0.05, 0.08, 0.03, 0.10],     # Asset 2
                         [0.12, 0.15, -0.05, 0.20],    # Asset 3
                         [0.03, 0.05, 0.02, 0.08]])    # Asset 4

print(f"\nAsset returns shape: {asset_returns.shape}")
print(f"Weights shape: {weights.shape}")

# Calculate portfolio returns using broadcasting
portfolio_returns = np.sum(asset_returns * weights.reshape(-1, 1), axis=0)
print(f"Portfolio quarterly returns: {portfolio_returns}")
```

```python
portfolios = np.array([[0.6, 0.3, 0.1],    # Conservative
                      [0.4, 0.4, 0.2],     # Balanced  
                      [0.2, 0.3, 0.5]])    # Aggressive

asset_performance = np.array([0.05, 0.10, 0.15])  # Expected returns

# Calculate expected returns for each portfolio using broadcasting
expected_returns = np.sum(portfolios * asset_performance, axis=1)
print(f"Portfolio allocations:\n{portfolios}")
print(f"Asset expected returns: {asset_performance}")
print(f"Portfolio expected returns: {expected_returns}")
```
